# start Exploratory Data analysis
price distribution

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df['price'], bins=30, kde=True)
plt.title('Price Distribution')
plt.show()
```

 Price Distribution

The Airbnb listings are spread across a wide range of prices, offering options for different customer budgets. Most listings are concentrated in the mid-price segment, indicating balanced pricing in the dataset.

# room type distribution

```python
sns.countplot(x='room_type', data=df)
plt.xticks(rotation=45)
plt.title('Room Type Distribution')
plt.show()
```

Entire apartments dominate the Airbnb market, while hotel rooms have the lowest availability.

# neighbourhood group distribution

```python
df['neighbourhood_group'].unique()
```

```python
df['neighbourhood_group'] = df['neighbourhood_group'].replace({
    'Brookln': 'Brooklyn'
})
```

```python
df['neighbourhood_group'] = df['neighbourhood_group'].replace({
    'brookln': 'Brooklyn',
    'manhattan': 'Manhattan',
    'manhatan':'Manhattan'
})
```

```python
sns.countplot(x='neighbourhood_group', data=df)
plt.xlabel('x-axis')
plt.title('Listings by Neighbourhood Group')
plt.show()
```

Manhattan and Brooklyn have the highest number of Airbnb listings, while Queens, the Bronx, and Staten Island have comparatively fewer listings.

# price vs Room type

```python
sns.boxplot(x='room_type', y='price', data=df)
plt.xticks(rotation=45)
plt.title('Price by Room Type')
plt.show()
```

Shared rooms and hotel rooms have slightly higher median prices, while overall price ranges are similar across all room types.

# price vs neighbourhood group

```python
sns.boxplot(x='neighbourhood_group', y='price', data=df)
plt.xticks(rotation=45)
plt.title('Price by Neighbourhood Group')
plt.show()
```

Prices are broadly similar across major neighbourhoods, with median prices around $620–$630 but considerable variation within each group.

# correlation heatmap

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

The heatmap shows number_of_reviews and reviews_per_month have the strongest correlation (0.60), while most other variables have very weak relationships.
